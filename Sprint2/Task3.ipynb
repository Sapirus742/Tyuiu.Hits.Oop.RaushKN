<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи: 
Создать базовый класс Review в C#, который будет представлять отзывы о продуктах или услугах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма. 

Требования к базовому классу Review: 

    • Атрибуты: ID отзыва (ReviewId), Текст отзыва (Text), Рейтинг (Rating). 
    • Методы: 
        o DisplayReview(): метод для отображения отзыва. 
        o RateProduct(): метод для присвоения рейтинга продукту. 
        o GetReviewDetails(): метод для получения деталей отзыва. 

Требования к производным классам: 
1. Отзыв Обслуживания (ServiceReview): Должен содержать дополнительные атрибуты, такие как Дата      посещения (VisitDate). Метод DisplayReview() должен быть переопределен для включения даты посещения в отображение отзыва. 
2. Отзыв Товара (ProductReview): Должен содержать дополнительные атрибуты, такие как Идентификатор продукта (ProductId). Метод RateProduct() должен быть переопределен для связывания рейтинга с конкретным продуктом. 
3. Отзыв Услуги (ServiceReview) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Время начала услуги (StartTime). Метод GetReviewDetails() должен быть переопределен для отображения времени начала услуги вместе с другими деталями отзыва. 

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;

public abstract class Review
{
    private int _reviewId;
    private string _text;
    private int _rating;

    public int ReviewId
    {   
        get {return _reviewId; } 
        set 
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("ID отзыва должен быть положительным числом");
            _reviewId= value;
        } 
    }
    public string Text 
    {   
        get {return _text; } 
        set 
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentNullException("Текст отзыва не может быть пустым");
            _text = value;
        } 
    }
    public int Rating 
    { 
        get { return _rating; }
        set
        {
            if (value < 0 || value > 5) throw new ArgumentOutOfRangeException("Оценка должна быть в диапазоне от 1 до 5");
            else _rating = value;

        } 
    }

    public static int TotalReview { get; private set; } = 0;

    public Review() { TotalReview++; }

    public static void DisplayTotalReview()
    {
        Console.WriteLine($"Общее количество отзывов: {TotalReview}");
    }

    public Review(int reviewId, string text, int rating)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = rating;
        TotalReview++;
    }

    public virtual void DisplayReview()
    {
        Console.WriteLine($"ID отзыва: {ReviewId}");
        Console.WriteLine($"Рейтинг: {Rating} из 5");
        Console.WriteLine($"Текст отзыва: {Text}");
    }

    public virtual void RateProduct(int newRating)
    {
        _rating = newRating;
    }

    public virtual void UpdateRating(int newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг отзыва {ReviewId} обновлен: {Rating} из 5");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID отзыва: {ReviewId}, Рейтинг: {Rating} из 5, Текст отзыва: {Text}";
    }

}

public interface IMediaContent
{
    bool HasMedia { get; set; }
    void AddMedia();
    void RemoveMedia();
    string GetMediaInfo();
}

public interface IRecommendationReview
{
    bool IsRecommended {get; set;}
    string RecommendationReason {get; set;}
    string GetRecommendationStatus();
}

public class ServiceReview : Review
{
    private DateTime _visitDate;

    public DateTime VisitDate 
    { 
        get {return _visitDate; } 
        set 
        {
            if (value > DateTime.Now)
                throw new ArgumentException("Дата посещения не может быть в будущем");
            _visitDate = value;
        }
    }

    public ServiceReview(int reviewId, string text, int rating, DateTime visitDate)
        : base(reviewId, text, rating)
    {
        VisitDate = visitDate;
    }

    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Дата посещения: {VisitDate:dd.MM.yyyy}\n");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", Дата посещения: {VisitDate:dd.MM.yyyy}";
    }
}

public class ProductReview : Review
{
    private int _productId;

    public int ProductID 
    { 
        get { return _productId; } 
        set 
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("ID продукта должен быть положительным числом");
            _productId = value;
        } 
    }

    public static int TotalProduct { get; private set; } = 0;

    public ProductReview() {TotalProduct++;}

    public static void DisplayTotalProduct()
    {
        Console.WriteLine($"Общее количество товаров: {TotalProduct}");
    }

    public ProductReview(int reviewId, string text, int rating, int productId)
        : base(reviewId, text, rating)
    {
        ProductID = productId;
        TotalProduct++;
    }

     public override void DisplayReview()
    {
        Console.WriteLine($"ID товара: {ProductID}");
        Console.WriteLine($"ID отзыва на товар: {ReviewId}");
        Console.WriteLine($"Рейтинг: {Rating} из 5");
        Console.WriteLine($"Текст отзыва: {Text}\n");
    }

    public override void UpdateRating(int newRating)
    {
        
        base.UpdateRating(newRating);
        Console.WriteLine($"Рейтинг для товара {ProductID} обновлен");
        
    }

    public override string GetReviewDetails()
    {
        return $"\nID отзыва: {ReviewId}, ID товара: {ProductID}, Рейтинг: {Rating} из 5, Текст отзыва: {Text}";
    }
}

public class ProductOverview : ProductReview, IMediaContent
{
    private string _advantages { get; set; }
    private string _disadvantages { get; set; }

    public string Advantages 
    {
        get 
        {
            if(string.IsNullOrEmpty(_advantages)) return "-";
            else return _advantages;
        }
        set {_advantages = value;}
    }

    public string Disadvantages
    {
        get
        {
            if(string.IsNullOrEmpty(_disadvantages)) return "-";
            else return _disadvantages;
        }
        set {_disadvantages = value;}
    }

    public bool HasMedia {get; set;}
    public int MediaCount {get; private set;}

    public ProductOverview(int reviewId, string text, int rating, int productId, string advantages = "", string disadvantages = "") : 
        base(reviewId, text, rating, productId)
    {
        Advantages = advantages;
        Disadvantages = disadvantages;
        HasMedia = false;
        MediaCount = 0;
    }

    public override void DisplayReview()
    {
        Console.WriteLine($"ID товара: {ProductID}");
        Console.WriteLine($"ID обзора на товар: {ReviewId}");
        Console.WriteLine($"Рейтинг: {Rating} из 5");
        Console.WriteLine($"Текст обзора: {Text}");
        Console.WriteLine($"Плюсы товара: {Advantages}");
        Console.WriteLine($"Минусы товара: {Disadvantages}\n");
    }

    public void AddMedia()
    {
        HasMedia = true;
        MediaCount++;
        Console.WriteLine($"\nМедиа контент добавлен к обзору {ReviewId}");
    }

    public void RemoveMedia()
    {
        if (MediaCount > 0)
        {
            MediaCount--;
            if(MediaCount == 0)
            {
                HasMedia = false;
            }
            Console.WriteLine($"Медиа контент удален из обзора {ReviewId}");
        }
    }

    public string GetMediaInfo()
    {
        if (!HasMedia) return "Медиа контент отсутствует";
        return $"Количество медиа: {MediaCount} файлов";
    }

    public override string GetReviewDetails()
    {
        return $"\nID отзыва: {ReviewId}, ID товара: {ProductID}, Рейтинг: {Rating} из 5, Текст отзыва: {Text}, Плюсы товара: {Advantages}, Минусы товара: {Disadvantages}";
    }
}

public class Service_Review : Review, IMediaContent, IRecommendationReview
{
    private DateTime _startTime;

    public DateTime StartTime 
    { 
        get { return _startTime; } 
        set 
        {
            if (value.TimeOfDay < new TimeSpan(8, 0, 0) || value.TimeOfDay > new TimeSpan(20, 0, 0))
                throw new ArgumentException("Время услуги должно быть между 8:00 и 20:00");
            _startTime = value;
        } 
    }
    
    public bool HasMedia {get; set;}
    public int MediaCount {get; private set;}
    public bool IsRecommended {get; set;}
    public string RecommendationReason {get; set;}

    public Service_Review( int reviewId, string text, int rating, DateTime startTime, bool isRecommended, string recommendationReason = "")
        : base(reviewId, text, rating)
    {
        StartTime = startTime;
        HasMedia = false;
        MediaCount = 0;
        IsRecommended = isRecommended;
        RecommendationReason = recommendationReason;
    }

    public override void DisplayReview()
    {
        Console.WriteLine($"ID отзыва на услугу: {ReviewId}");
        Console.WriteLine($"Время начала услуги: {StartTime:HH:mm}");
        Console.WriteLine($"Рейтинг: {Rating} из 5");
        Console.WriteLine($"Текст отзыва: {Text}");
        Console.WriteLine(IsRecommended ? $"Рекомендую : {RecommendationReason}\n" : "Не рекомендую\n");
    }

    public void AddMedia()
    {
        HasMedia = true;
        MediaCount++;
        Console.WriteLine($"\nМедиа контент добавлен к обзору {ReviewId}");
    }

    public void RemoveMedia()
    {
        if (MediaCount > 0)
        {
            MediaCount--;
            if(MediaCount == 0)
            {
                HasMedia = false;
            }
            Console.WriteLine($"Медиа контент удален из обзора {ReviewId}");
        }
    }

    public string GetMediaInfo()
    {
        if (!HasMedia) return "Медиа контент отсутствует";
        return $"Количество медиа: {MediaCount} файлов";
    }

    public string GetRecommendationStatus()
    {
        return IsRecommended ? $"Рекомендую: {RecommendationReason}" : "Не рекомендую";
    }

    public override string GetReviewDetails()
    {
        return $"\nID отзыва: {ReviewId}, Рейтинг: {Rating} из 5, Текст отзыва: {Text}, Время начала услуги: {StartTime:HH:mm}, " + $"{(IsRecommended ? $"Рекомендую: {RecommendationReason}" : "Не рекомендую")}";
    }
}

ServiceReview one = new ServiceReview(1, "Хороший отзыв", 4, new DateTime(1997, 1, 15));
ProductReview two = new ProductReview(2, "Нормальный отзыв", 3, 1);
Service_Review three = new Service_Review(3, "Плохой отзыв", 2, new DateTime(2024, 1, 15, 12, 14, 0), false);
ProductOverview four = new ProductOverview(4, "Средненький товар", 3, 1, "Хорошая цена и стильный дизайн", "Прослужил не долго");
one.DisplayReview();
two.DisplayReview();
three.DisplayReview();
four.DisplayReview();
ServiceReview.DisplayTotalReview();
ProductReview.DisplayTotalProduct();

four.AddMedia();
Console.WriteLine(four.GetMediaInfo());
four.RemoveMedia();
Console.WriteLine(four.GetMediaInfo());
```
