```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Blossom()
    {
        Console.WriteLine($"{Name} цветёт");
    }

    public virtual void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} вырос на {growth} см. Новая высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, string color, int petalCount, float height) 
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Blossom()
    {
        Console.WriteLine($"{Type} {Name} распускается и радует глаз {Color} лепестками!");
    }

    public void Wilting()
    {
        Console.WriteLine($"{Type} {Name} начинает увядать и его {PetalCount} лепесков (-а) опадают...\n");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, int age, string leafType, float height) 
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Blossom()
    {
        Console.WriteLine($"{Type} {Name} покрывается новыми {LeafType} листьями");
    }

    public void GrowAge(int years)
    {
        Age += years;
        Console.WriteLine($"{Type} {Name} стало старше на {years} лет. Новый возраст: {Age} лет");
    }
}

Flower myFlower = new Flower("Тюльпан", "красными", 5, 10);
Tree myTree = new Tree("Дуб", 25 , "лиственными", 250);
myFlower.Blossom();
myFlower.Grow(10);
myFlower.Wilting();

myTree.Blossom();
myTree.Grow(20);
myTree.GrowAge(12);
```
