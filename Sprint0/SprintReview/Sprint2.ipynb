```polyglot-notebook
public class Employee
{
    private string _name;
    private string _position;
    private double _salary;

    public string Name
    {
        get { return _name;}
        set { _name = value;}
    }

    public string Position
    {
        get { return _position;}
        set { _position = value;}
    }

    public double Salary
    {
        get { return _salary;}
        set { _salary = value;}
    }

    public Employee (string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"\nИмя сотрудника: {Name}, Должность сотрудника: {Position}, Зарплата сотрудника: {Salary}");
    }
        
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers(); 
}

public class Manager : Employee
{
    private string _department;

    public string Department
    {
        get { return _department;}
        set { _department = value;}
    }

    public Manager (string name, string position, double salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.Write($"Отдел, которым управляет менеджер: {Department}\n");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    private string _projectName;
    private List<string> teamMembers;

    public string ProjectName
    {
        get { return _projectName;}
        set { _projectName = value;}
    }

    public ProjectManager (string name, string position, double salary, string department, string projectName) : base(name, position, salary, department)
    {
        ProjectName = projectName;
        teamMembers = new List<string>();
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.Write($"Название проекта: {ProjectName}\n");
    }

    public void AddTeamMember(string memberName)
    {
        teamMembers.Add(memberName);
    }

    public string GenerateReport()
    {
        return $"\nОтчет по проекту: \nНазвание проекта: {ProjectName}\n" +
               $"Руководитель: {Name}\nОтдел: {Department}\nКоличество членов команды: {teamMembers.Count}\n";
    }

    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }
}

Employee myEmployee = new Employee("Николай", "Основной сотрудник компании", 65000.35);
Manager myManager = new Manager("Олег", "Главный менеджер", 85000, "Перспективные разработоки");
ProjectManager myProjectManager = new ProjectManager("Павел", "Основной сотрудник компании", 75000.89, "Новейшие разработки", "Система новейшей поддержки клиентов");

myEmployee.PrintInfo();
myManager.PrintInfo();
myProjectManager.PrintInfo();

myProjectManager.AddTeamMember("Семен");
myProjectManager.AddTeamMember("Толик");

Console.WriteLine(myProjectManager.GenerateReport());
Console.Write($"Члены проекта {myProjectManager.ProjectName}:");
foreach(var member in myProjectManager.GetTeamMembers())
{
    Console.Write($" {member}");
}
```
