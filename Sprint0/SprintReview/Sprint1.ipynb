```polyglot-notebook
public class Product 
{
    private string _name;
    private decimal _price;
    private int _quantity;
    protected string _category;
    internal string _description;

    public string Name 
    {
        get { return _name; }
        set { _name = value; }
    }
    public decimal Price
    {
        get { return _price; }
        set { _price = value; }
    }
    public int Quantity 
    {
        get { return _quantity; }
        set { _quantity = value; }
    }
    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }
    public string Description
    {
        get { return _description; }
        set { _description = value; }
    }

    public Product(string name, decimal price, int quanity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quanity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return _price * _quantity;
    }

    public void UpdateQuanity(int amount)
    {
        _quantity += amount;
        if (_quantity < 0) _quantity = 0;
    }

    public string GetProductInfo()
    {
        return $"Товар: {_name}, Цена: {_price:C}, Количество: {_quantity}, Категория: {_category}, Описание: {_description}";
    }
}

public class Seller 
{
    private string _name;
    private string _employerId;
    protected decimal _salary;
    internal string _contactInfo;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }
    public string EmployerId
    {
        get { return _employerId; }
        set { _employerId = value; }
    }
    public decimal Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }
    public string ContactInfo
    {
        get { return _contactInfo; }
        set { _contactInfo = value; }
    }

    public Seller(string name, string employerId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployerId = employerId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void SellProduct(Product product, int quanity)
    {
        if (product.Quantity >= quanity)
        {
            product.UpdateQuanity(-quanity);
            Console.WriteLine($"Продано {quanity} едениц товара '{product.Name}' продавцом {_name}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара '{product.Name}' для продажи. Доступно: {product.Quantity}");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {_name}, ID: {_employerId}, Зарплата: {_salary:C}, Контакты: {_contactInfo}";
    }

}

public class Store
{
    private string _storeName;
    private string _location;
    public string storeHours;
    private List<Seller> _sellers;
    private List<Product> _products;

    public string StoreName
    {
        get { return _storeName; }
        set { _storeName = value; }
    }
    public string Location
    {
        get { return _location; }
        set { _location = value; }
    } 

    public Store(string storeName, string location, string stireHours)
    {
        _storeName = storeName;
        _location = location;
        this.storeHours = storeHours;
        _sellers = new List<Seller>();
        _products = new List<Product>();
    }

    public void AddSeller(Seller seller)
    {
        _sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин {_storeName}");
    }

    public void AddProduct(Product product)
    {
        _products.Add(product);
        Console.WriteLine($"Товар '{product.Name}' добавлен магазину {_storeName}");
    }

    public void ListProducts()
    {
        Console.WriteLine($"\nВсе товары в магазине {_storeName}:");
        foreach (var product in _products)
        {
            Console.WriteLine($"- {product.GetProductInfo()}");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {_storeName}, Адрес: {_location}, Часы работы: {storeHours}, Продавцов: {_sellers.Count}, Товаров: {_products.Count}";
    }
}

Product pc = new Product("Компьютер", 5000m, 15, "Электроника", "Настольное устройство, помогающее в разных делах");
Product ruphone = new Product("Смартфон", 450m, 10, "Электроника", "Небольшое подручное устройство, способное улучшить жизнь");
Product shovel = new Product("Лопата", 20m, 50, "Хозяйственное оборудование", "Ручное приспособление для копания");
Seller seller1 = new Seller("Сидорович", "1", 100m, "Номер телефона: +7890...");
Seller seller2 = new Seller("Продавец2", "2", 150m, "Номер телефона: +7896...");
Store store = new Store("Мвидео", "Улица: ... , дом: ...", "8:00 - 21:00");

Console.WriteLine(pc.GetTotalPrice());
pc.UpdateQuanity(-10);
Console.WriteLine(pc.Quantity);
Console.WriteLine(pc.GetProductInfo());

store.AddSeller(seller1);
store.AddProduct(ruphone);

seller1.SellProduct(ruphone, 5);
Console.WriteLine(seller2.GetSellerInfo());

store.ListProducts();
Console.WriteLine(strore.GetStoreInfo());
```
